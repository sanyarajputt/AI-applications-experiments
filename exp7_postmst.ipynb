**EXP-7**
Fitts law - To verify Fitts' Law using Python by computing the index of difficulty for several target conditions and fitting movement time against it.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Distance (D) and Width (W)
D = np.array([100, 200, 300, 400, 500])
W = np.array([50, 50, 50, 50, 50])
```

```python
ID = np.log2(2 * D / W)

# Simulated Movement Time (MT)
# MT = a + b * ID + small noise
a = 0.2
b = 0.1

MT = a + b * ID + np.random.normal(0, 0.01, len(ID))
```

```python
ID_reshaped = ID.reshape(-1, 1)

model = LinearRegression()
model.fit(ID_reshaped, MT)

MT_pred = model.predict(ID_reshaped)
```

```python
print("Index of Difficulty:", ID)
print("Movement Time:", MT)
```

```python
plt.figure()
plt.scatter(ID, MT, label="Observed Data")
plt.plot(ID, MT_pred, label="Fitted Line")

plt.xlabel("Index of Difficulty (ID)")
plt.ylabel("Movement Time (MT)")
plt.title("Fitts' Law Verification")

plt.legend()
plt.show()
```
