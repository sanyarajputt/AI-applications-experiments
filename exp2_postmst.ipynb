```python
import random
import hashlib
import matplotlib.pyplot as plt

```

```python
def get_sensor_data():
    return {
        "traffic": random.randint(0, 100),
        "pollution": random.randint(0, 100),
        "energy": random.randint(0, 100)
    }
    #iot stimulation
```

```python
def ai_decision(data):
    decisions = []

    if data["traffic"] > 70:
        decisions.append("Reduce Traffic Flow")

    if data["pollution"] > 60:
        decisions.append("Pollution Alert")

    if data["energy"] > 80:
        decisions.append("Activate Backup Power")

    if not decisions:
        decisions.append("All Normal")

    return decisions
#ai decision system
```

```python
blockchain = []

def create_block(data, decision, prev_hash):
    block_string = str(data) + str(decision) + prev_hash
    block_hash = hashlib.sha256(block_string.encode()).hexdigest()

    return {
        "data": data,
        "decision": decision,
        "prev_hash": prev_hash,
        "hash": block_hash
    }
```

```python
traffic_list = []
pollution_list = []
energy_list = []

prev_hash = "0"

for i in range(10):
    data = get_sensor_data()
    decision = ai_decision(data)

    block = create_block(data, decision, prev_hash)
    blockchain.append(block)

    prev_hash = block["hash"]

    # Store for plotting
    traffic_list.append(data["traffic"])
    pollution_list.append(data["pollution"])
    energy_list.append(data["energy"])

    print(f"\nStep {i+1}")
    print("Sensor Data:", data)
    print("AI Decision:", decision)
    print("Block Hash:", block["hash"][:10], "...")
```

```python
plt.figure()
plt.plot(traffic_list, label="Traffic")
plt.plot(pollution_list, label="Pollution")
plt.plot(energy_list, label="Energy")

plt.title("Smart City Sensor Data")
plt.xlabel("Time Step")
plt.ylabel("Values")
plt.legend()
plt.show()
```
